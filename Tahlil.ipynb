```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("dataset.csv")
df.head(10)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["customer_rating"]=df["customer_rating"].fillna(df["customer_rating"].mean())
df
```

```python
df["competitor_price_index"]=df["competitor_price_index"].fillna(df["competitor_price_index"].mean())
df
```

```python
df["promo_active"].mode()
```

```python
df["promo_active"]=df["promo_active"].fillna(df["promo_active"].mode())
df
```

```python
df.columns
```

```python
df.dtypes
```

```python
corr = df[["year","month","avg_unit_price","units_sold","returns", "discount_pct", "customer_rating", "competitor_price_index", "marketing_spend", "monthly_revenue"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1, center=0)
plt.show()
```

```python
plt.figure(figsize=(15,6))
x = df["category"].unique()
y = df.groupby("category")["customer_rating"].sum()
plt.title("Mahsulotlarni reytingi.",size=20)
plt.xlabel("Kategoriya",size=16)
plt.ylabel("Reyting",size=16)
plt.grid(color='orange', linestyle='-', linewidth=0.5, alpha=0.4)
bars = plt.bar(x, y, align='center', color='blue', edgecolor='black', alpha=1)

plt.bar_label(bars, labels=[f"{i:.2f}%" for i in y], padding=3, fontsize=12, fontweight='bold')

plt.show()
```

```python
plt.figure(figsize=(15,6))
x = df["category"].unique()
y = df.groupby("category")["monthly_revenue"].sum()/sum(df["monthly_revenue"])*100
plt.title("Mahsulotlarni foyda ulushi foizda.",size=20)
plt.xlabel("Kategoriya",size=16)
plt.xlabel("Kategoriya",size=16)
plt.ylabel("Foyda")
plt.grid(color='orange', linestyle='-', linewidth=0.5, alpha=0.4)
bars = plt.bar(x , y, align='center', color='blue', edgecolor='black', alpha=1)

plt.bar_label(bars, labels=[f"{i:.2f}%" for i in y], padding=3, fontsize=12, fontweight='bold')

plt.show()
```

```python
plt.figure(figsize=(10, 8))
x = df["region"].unique()
y = df.groupby("region")["monthly_revenue"].count()
plt.title("Regionlar bo'yicha umumiy foyda ulushi.")
plt.pie(y, labels=x, autopct='%1.1f%%', startangle=90)
plt.show()
```
